## Exercise 4 - Softmax and loss

Given logits `[2.0, 1.0, 0.0]` and target class `0`, compute:

- the softmax probabilities (approximate values are fine), and
- the negative log likelihood for the target class.

You can use the cell below to compute it numerically and check your by-hand answer.

````python
import math

logits = [2.0, 1.0, 0.0]
exps = [math.exp(x) for x in logits]
total = sum(exps)
probs = [e / total for e in exps]
target = 0
nll = -math.log(probs[target])

print("probs:", [round(p, 4) for p in probs])
print(f"nll for target={target}: {nll:.4f}")
````
